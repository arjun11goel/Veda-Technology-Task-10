# The Relationship Between AI, ML, and Deep Learning
**Artificial Intelligence (AI)** is the broadest concept. It encompasses the entire field of computer science dedicated to creating machines capable of performing tasks that typically require human intelligence (such as reasoning, visual perception, and decision-making).

**Machine Learning (ML)** is a specific subset of AI. Instead of being explicitly programmed with hardcoded rules, ML systems use algorithms to identify patterns in data, allowing them to "learn" and improve their performance over time.

**Deep Learning (DL)** is a specialised subset of Machine Learning. It utilises multi-layered artificial neural networks (inspired by the human brain) to process highly complex data. While traditional ML often requires human intervention to extract features, Deep Learning automatically discovers the representations needed for detection or classification from raw data.

| Concept | Definition | Key Difference | Real-World Examples|
| :--- | :--- | :--- | :--- |
| **AI** | Systems designed to mimic human cognitive functions and intelligence.| The overarching category; can include simple rule-based logic or complex learning algorithms. | Chess-playing computer programs (using search trees) or rule-based chatbots. |
| **ML** | Algorithms that parse data, learn from it, and make informed decisions based on what they've learned. | Requires structured data and often relies on human-guided feature engineering. | Netflix/Spotify recommendation systems predicting what you want to watch or hear. |
| **DL** | Neural networks with three or more layers that attempt to simulate human learning to process unstructured data. | Automatically performs feature extraction and handles massive, unstructured datasets. | Image recognition (e.g., Apple's FaceID) and self-driving car vision systems. |


# Real-World Examples Expanded
**AI Example:** A smart home thermostat that adjusts temperature based on a schedule and basic "if/then" rules programmed by the user.

**ML Example:** An email spam filter that looks at thousands of emails categorized as "spam" or "not spam," learning to identify key words (like "lottery" or "urgent") to automatically filter future messages.

**Deep Learning Example:** Medical imaging software that analyzes thousands of raw MRI scans using deep neural networks to accurately detect early signs of tumors without a radiologist manually pointing out the features of a tumor first.

# Interview Questions & Answers
**1. What is Artificial Intelligence?**
Artificial Intelligence is a broad field of computer science focused on building smart machines capable of performing tasks that typically require human intelligence. This includes capabilities like reasoning, problem-solving, understanding natural language, and perceiving environments.

**2. What is Machine Learning?**
Machine Learning is a subset of AI that focuses on the development of algorithms that allow computers to learn from and make predictions or decisions based on data. Instead of being explicitly programmed to perform a task, ML models adjust their internal parameters through exposure to training data.

**3. What is the difference between Machine Learning and Deep Learning?**
The main differences lie in data handling and feature extraction. Traditional Machine Learning usually requires manual "feature engineering" (a human deciding which variables are important) and performs well on structured data. Deep Learning uses artificial neural networks with multiple layers to automatically extract features from raw, unstructured data (like images, audio, or text). Furthermore, Deep Learning models typically require significantly more computing power and massive amounts of data to outperform traditional ML algorithms.
